## python 类和对象

```python
# 创建一个对象, 并使用对象
class People:
    def __init__(self, id, name, age):
        print('id       = ' + str(id))
        print('name     = ' + str(name))
        print('age      = ' + str(age))

    def _sayHello(self, param):
        print('Hello ' + str(param))

zhangsan = People(1, 'zhangsan', 18)
zhangsan._sayHello('zhangsan')
    
```

```python
# python 中的继承相关的知识
class Parent: 
    def __init__(self):
        self.__state = 'parent_state'
        print('parent __init__ method invoke')
    def show(self):
        print(self.__state)

class Child(Parent): 
    def __init__(self):
        super().__init__()
        self.__state = 'child'

child = Child()
# 继承中的坑, 使用 __paramName 会 自动变成 _className__paramName 的自动变更名称的约定
print(child._Child__state)
        
```
