## 自动求导和张量

### 1. 张量

```python
import torch

# 创建一个张量, 开启张量的自动求导追踪
x = torch.ones(2, 2, requires_grad=True)
print(x)
```

```python
# 对张量 x 进行操作；
y = x + 2
print(y)
```

```python
print(y.grad_fn)
```

```python
# 对张量 y 再进行一个操作 
z = y * y * 3
out = z.mean()
print(z)
print(z.size)
print(out)
```

```python
# 可以手动对自动求导进行打开; 默认的是关闭的
a = torch.randn(2, 2)
a = ((a * 3) / (a - 1) )
print(a.requires_grad)
a.requires_grad_(True)
print(a.requires_grad)
b = (a * a).sum()
print(b.grad_fn)
```

### 2.梯度

```python
# 为了查看 梯度; 这块将 x,y,z,out 都打印出来看看
print(x)
x.retain_grad()
print(y)
y.retain_grad()
print(z)
z.retain_grad()
print(out)
```

```python
# 使用 backward() 进行 反向传播
out.backward()
```

```python
# 只有叶子节点 的 grad 有值; 这个 grad 值; 是下一个函数对当前 张量的求导
print(x.grad)
print(y.grad)
print(z.grad)
print(out.grad)
# is_leaf 是叶子节点; grad_fn: 查看是不是不是 计算出来的
print(x.is_leaf)
print(y.grad_fn)
```

```python
print(x.dtype, y.dtype, z.dtype, out.dtype)
```

```python
# 看一下个 雅可比矩阵的 案例
x = torch.randn(3, requires_grad = True)
print(x)
```

```python
y = x * 2
print(y)
print(y.data.norm())

```
