## 自动求导和张量

### 1. 张量

```python
import torch

# 创建一个张量, 开启张量的自动求导追踪
x = torch.ones(2, 2, requires_grad=True)
print(x)
```

```python
# 对张量 x 进行操作；
y = x + 2
print(y)
```

```python
print(y.grad_fn)
```

```python
# 对张量 y 再进行一个操作 
z = y * y * 3
out = z.mean()
print(z)
print(z.size)
print(out)
```

```python
# 可以手动对自动求导进行打开; 默认的是关闭的
a = torch.randn(2, 2)
a = ((a * 3) / (a - 1) )
print(a.requires_grad)
a.requires_grad_(True)
print(a.requires_grad)
b = (a * a).sum()
print(b.grad_fn)
```

### 2.梯度

```python
# 为了查看 梯度; 这块将 x,y,z,out 都打印出来看看
print(x)
x.retain_grad()
print(y)
y.retain_grad()
print(z)
z.retain_grad()
print(out)
```

```python
# 使用 backward() 进行 反向传播
out.backward()
```

```python
# 反向传播学习 backward()
x = torch.randint(1, 10, size = (3,), dtype = float)
print(x, x.requires_grad, x.is_leaf)

x.requires_grad_()
y = x * 2

while y.data.norm() > 1000:
    y = y * y * 2

y.retain_grad()

z = y * y * y * 3
z.retain_grad()

gradients = torch.tensor([1.0, 1.0, 1.0])
z.backward(gradients)

print('x = ' + str(x), 'x.grad = ' + str(x.grad))
print('y = ' + str(y), 'y.grad = ' + str(y.grad))
print('z = ' + str(z), 'z.grad = ' + str(z.grad))
```
