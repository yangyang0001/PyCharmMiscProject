## 张量相关

```python
import torch

# 创建一个 5 * 3矩阵; 没有初始化的 张量
x = torch.empty(5 * 3)
print(x)
```

```python
# 创建一个 5 * 3的随机 张量
x = torch.rand(5 * 3)
print(x)
```

```python
# 创建一个数据类型为 long; 5 * 3 的张量
x = torch.zeros(5, 3, dtype = torch.long)
print(x)
```

```python
# 使用 tensors 现有数据 创建 张量
x = torch.tensor([5.5, 3])
print(x)
```

```python
# 使用已经存在的张量 创建张量
x = x.new_ones(5, 3, dtype = torch.long)
print(x)

# 使用 like 系列
x = torch.rand_like(x, dtype = torch.float)
print(x)
```

```python
# 看下 x.size() 通过提示 可以看出来是 function
print(x.size())
```

```python
# 张量的 + 法
y = torch.rand(size = (5, 3), dtype = torch.float)
display(x, y)
print(x + y)
```

```python
# 使用 torch.add 方法进行 +
print(torch.add(x, y))
```

```python
# 使用 result 作为输出
result = torch.empty(5, 3)
torch.add(x, y, out = result)
print(result)
```

```python
# 将 x 添加到 y 上; 改变y的值
y.add_(x)
print(x)
print(y)
```
