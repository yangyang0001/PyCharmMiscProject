## 张量相关

### 1.开始

```python
import torch

# 创建一个 5 * 3矩阵; 没有初始化的 张量
x = torch.empty(5 * 3)
print(x)
```

```python
# 创建一个 5 * 3的随机 张量
x = torch.rand(5 * 3)
print(x)
```

```python
# 创建一个数据类型为 long; 5 * 3 的张量
x = torch.zeros(5, 3, dtype = torch.long)
print(x)
```

```python
# 使用 tensors 现有数据 创建 张量
x = torch.tensor([5.5, 3])
print(x)
```

```python
# 使用已经存在的张量 创建张量
x = x.new_ones(5, 3, dtype = torch.long)
print(x)

# 使用 like 系列
x = torch.rand_like(x, dtype = torch.float)
print(x)
```

```python
# 看下 x.size() 通过提示 可以看出来是 function
print(x.size())
```

```python
# 张量的 + 法
y = torch.rand(size = (5, 3), dtype = torch.float)
display(x, y)
print(x + y)
```

```python
# 使用 torch.add 方法进行 +
print(torch.add(x, y))
```

```python
# 使用 result 作为输出
result = torch.empty(5, 3)
torch.add(x, y, out = result)
print(result)
```

```python
# 将 x 添加到 y 上; 改变y的值
y.add_(x)
print(x)
print(y)
```

```python
# 分片的使用方法
print(x[:, 1])
```

```python
x = torch.randn(4, 4)
y = x.view(16)
# 这里的 -1, 表示不知道这个数不清楚; 让程序自己计算
z = y.view(-1, 8)
print(x, y, z)
print(x.size(), y.size(), z.size())
```

```python
# 生成只有一个元素的 一维张量
x = torch.randn(1)
print(x)
print(x.item())
```

### 2.numpy 转换

```python
a = torch.ones(5)
print(a)
```

```python
# 使用 numpy 产生的数组 和 原始a张量使用同一份数据; 张量 ---> 数组 的转化: 使用 a.numpy()
b = a.numpy()
print(b)
```

```python
b[1] = 10
print(a)
```

```python
a.add_(1)
print(a)
print(b)
```

```python
# numpy库 创建的数组 转换成 张量
import numpy as np
a = np.ones(5)
b = torch.from_numpy(a)
print(a, b)
```

```python
# 所有张量的计算都是基于 CPU 的; 字符串张量 不支持 cuda 张量
torch.cuda.is_available()
```

```python
torch.cuda.device_count()      # 可用 GPU 数量
torch.cuda.get_device_name(0)  # 第 0 张 GPU 的名称
torch.cuda.current_device()    # 当前默认 GPU 编号
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(device)

x = torch.tensor([1.0])
y = torch.ones_like(x, device=device)  # 直接在 GPU 上创建
x = x.to(device)                       # 把 x 搬到 GPU
z = x + y
print(z)                     # tensor([2.], device='cuda:0')
print(z.to("cpu", torch.double))  # tensor([2.], dtype=torch.float64)
```
