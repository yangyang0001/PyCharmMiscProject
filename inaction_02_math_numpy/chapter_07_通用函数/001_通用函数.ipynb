## 通用函数

### 1. 单个元素 常用的函数

abs、sqrt、square、exp、log、sin、cos、tan、maxinmum、minimum、                                             
all、any、inner、clip、round、trace、ceil、floor

```python
import numpy as np

a = np.array([-1, 3, 5, -2, 11, 10, -6], dtype=np.float16)
a
```

```python
# 使用绝对值函数 abs
np.abs(a)
```

```python
# 使用开平方函数 (张量中不能有负数)
np.sqrt(a)
```

```python
# 使用 平方函数
np.square(a)
```

```python
# exp 自然底数e 的多少次幂
np.exp(3)
```

```python
np.e**3
```

```python
# 求自然底数e的对数
np.log(20)
```

```python
# 三角函数
np.sin(np.pi / 2)
```

```python
np.cos(0)
```

```python
np.tan(np.pi / 4)
```

```python
# 单个元素取最大
np.maximum(a[0], a[1])
```

```python
# 打印一下a 
a
```

```python
c = np.random.randint(10, 17, size =7);
c
# 两个数组中 每个位置都选大的
np.maximum(a, c)
```

```python
# any 如果数据有一个不是 0 的就返回 True
nd1 = np.array([1, 3, 5, 0])
nd2 = np.array([-1, 13, 15, 10])
np.any(nd1) # True
np.any(nd2) # True
```

```python
# all 只有所有数据都不是 0 才返回 True
np.all(nd1)  #False
np.all(nd2)  #True
```

```python
# inner 返回的是两个数组的内积
a = np.array([1, 2, 3, 4, 5, 6])
b = np.array([1, 2, 3, 4, 5, 5])
np.inner(a, b)
```

```python
# clip 有裁剪的意思; 
nd1 = np.random.randint(0, 100, size = 30)
nd1

np.clip(nd1, 20, 30) # <20的变为20; 大于30的变为30

```

```python
# round 保留小数的位数
nd1 = np.random.rand(10, 5)
nd1
nd1.round(2)
```

```python
# ceil 天花板；向上取整
np.ceil(np.array([-1.5, 2.7, 3.1, 4.5]))
```

```python
# floor 地板; 向下取整
np.floor(np.array([-1.5, 2.7, 3.1, 4.5]))
```

```python
a = np.random.randint(1, 10, size = (3, 5))
a
```

```python
# 求的是 坐标 i = j 的所有值的和; a[0][0] + a[1][1] + a[2][2]
np.trace(a)
```
