## 复制和视图

### 1. 完全没有复制; 对同一份数据进行操作

```python
import numpy as np

nd1 = np.random.randint(1, 10, size = 3)
nd1 
```

```python
nd2 = nd1
```

```python
nd2 is nd1
```

### 2. 视图操作；叫法比较多 视图 或 查看 或 浅拷贝

```python
a = np.random.randint(1, 100, size = 5)
b = a.view()
display(a, b)
```

```python
# 证明 a 和 b 不是一回事
a is b
```

```python
# 查看 a 是否拥有自己的数据
a.flags.owndata
```

```python
# 查看 b 是否拥有自己的数据; 使用的数据还是 a 的数据
b.flags.owndata
```

```python
a[0] = 1024
b[0] = 1
display(a, b)
a is b
```

### 3. 深拷贝

```python
a = np.random.randint(1, 10, size = 5)
b = a.copy()
display(a, b)
```

```python
# 深copy 含义: 完全复制一份自己的数据
a is b
```

```python
a[0] = 1024
b[0] = 2048
display(a, b)
```

```python
# 那么 copy() 方法如何使用呢?
a = np.arange(1e8)
a
```

```python
# a的数据非常大, 只使用其中的一部分；复制出来后, 删除掉可以省很多内存
b = a[[1, 2, 4]].copy()
b
```
