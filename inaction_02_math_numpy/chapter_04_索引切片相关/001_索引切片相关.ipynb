## 索引和切片

### 1. 索引和切片基础知识

```python
import numpy as np

a = np.random.randint(1, 100, size = 10)
a
```

```python
# 获取一个元素
a[3]
```

```python
# 获取多个元素; 目前标准就是 [里面填写下标； 多个用逗号分割]
a[[1, 3, 5]]
```

```python
# 切片；一刀切下去得到一块东西的感觉 类似于 substring的感觉; 左闭右开; 0表示开始下标; 3表示一共获取几个元素
a[0:3]
```

```python
# 切片; 如果冒号前不写, 默认是从0开始； 冒号后面不写则默认 从开始元素到最后一个元素的个数
a[:3]
```

```python
# 切片; 如果冒号前不写, 默认是从0开始； 冒号后面不写则默认 从开始元素到最后一个元素的个数
a[5:]
```

```python
# 从某个索引开始获取
a[5:9]
```

```python
# 为了方便查看 打印一下 a
a
```

```python
# 从索引 = 0开始 每2个当中获取一个
a[::2]
```

```python
# 从索引 = 3开始 每3个取一个
a[3::3]
```

```python
# 数组 前后下标翻转; [意思为 从后往前选择]
a[::-1]
```

```python
# 数组 前后下标翻转; [意思为 从后往前选择] 两个里面取一个
a[::-2]
```

```python
# 数组
a[5::-3]
```

```python
# 打印一下数组方便查看
a
```

```python
# 索引从1开始到7结束; 每隔2个取一个 [正向获取 位置3>0; 位置1<位置2] [反向获取 位置3<0; 位置1 > 位置2]
a[1:7:2]
```

```python
a[5:2:-1]
```

### 2. 多维张量的切片

```python
b = np.random.randint(0, 30, size = (10, 5))
b
```

```python
# 多维张量切片; 获取整行
b[1]
```

```python
# 多维张量取 多行
b[[1, 3, 5]]
```

```python
# 取固定纬度的值; 用逗号分隔纬度 行=1 列=3
b[1, 3]
```

```python
# 取一行中多个列的值 行=1, 列=[1,2,3]
b[1,[1,2,3]]
```

```python
# 取多行 多列的数据
b[2::7]
```

```python
# 取多行 多列的数据; 行从2到最后每隔7个 取一个; 列从1到最后 每隔3个取一个
b[2::7, 1::3]
```

```python
# 取最后一个元素; -1表示倒着数
b[-1, -1]
```

```python
# 获取倒着数 一行数据 写法 和 b[-2, [-1,-2,-3]] 一样 
b[-2, (-1,-2,-3)]
```

### 3. 花式索引和索引技巧

```python
a = np.arange(20)
a
```

```python
# 切片时 不是深拷贝; 公用一份数据
b = a[3:7]
b
```

```python
b[0] = 100
display(a, b)
```

```python
a.flags.owndata
```

```python
b.flags.owndata
```

```python
a = np.arange(20, 100, 10)
a
```

```python
# 在填写索引的时候给了一个数组; 这种索引就是花式索引; 花式索引返回的数据 都是经过深拷贝的数据;独立于原始数据的新数据
b = a[[1, 3]]
b
```

```python
b.flags.owndata
```

```python
a = np.random.randint(0, 151, size = (100, 3))
a
```

```python
# 筛选出所有 元素>120 的元素
cond = a >= 120
```

```python
a[cond]
```

```python
# 筛选出 每个列都 > 120的
cond2 = cond[:,0] * cond[:,1] * cond[:,2]
```

```python
a[cond2]
```

```python
# 筛选出 一行中都>100 或者 一行中都<30
cond1 = a>100
cond2 = a<30
display(a[cond1[:, 0]*cond1[:, 1]*cond1[:, 2]], a[cond2[:, 0]*cond1[:, 1]*cond1[:, 2]])
```
