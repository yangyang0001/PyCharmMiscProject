## 数据类型

### 1. 常用数据类型

int8, int16, int32, int64 ..., uint8 无符号

float16, float32, float64 ...

str字符串类型

int8 表示能存储 256个有符号的 数字; -128到127 之间的数字

uint8表示能存储 256个无符号的 数字; 0到255 之间的数字

```python
import numpy as np

# 创建一个数组
nd1 = np.array([2, 3], dtype=np.int8)
display(nd1)
```

```python
# 注意下 类型能够包含的数字 最大和最小都是多少; 这块需要了解一下
np.array([-3, -5, 0, 255, 0, 256], dtype = np.uint8)
```

```python
np.random.randint(0, 100, size = 10, dtype = np.int8)
```

```python
nd = np.random.rand(10, 2)
nd
```

```python
nd.dtype
```

```python
# 产生一个新的数组; 原数组并没有改动
nd1 = np.asarray(nd, dtype ='float16')
nd1
```

```python
display(nd.dtype, nd1.dtype)
```

```python
nd1.astype(np.float16)
```

### 2. 数据或者矩阵的计算

### 2.1 基本运算

```python
nd1 = np.random.randint(1, 100, size = 5)
nd2 = np.random.randint(1, 10, size = 5)
display(nd1, nd2)
```

```python
# 逐个元素相加
nd1 + nd2
```

```python
# 逐个元素相减
nd1 - nd2
```

```python
# 逐个元素相乘
nd1 * nd2
```

```python
# 逐个元素相除
nd1 / nd2
```

```python
# 幂等运算
nd1 ** nd2
```

```python
# 幂等运算 power 函数, 内部传递两个参数
np.power(nd1, nd2)
```

```python
# 对数函数
np.log(10)
```

```python
np.log10(1000)
```

```python
np.log2(1024)
```

### 2.2 逻辑运算

```python
display(nd1, nd2)
```

```python
# 问一下 nd1 和 nd2 数组中每个对应的元素 是否都 > 呢
nd1 > nd2
```

```python
nd1 < nd2
```

```python
nd1 >= nd2
```

```python
nd1 == nd2
```

### 2.3 数组和标量的计算

```python
display(nd1, nd2)
```

```python
# 数组中每个元素都 + 10
nd1 + 10
```

```python
# 数组中每个元素都 - 10
nd1 - 10
```

```python
# 数组中每个元素都 * 10
nd1 * 10
```

```python
# 数组中每个元素都 / 10
nd1 / 10
```

```python
# 数组也可以做分母；0不能做分母, 否则报出错误 inf
2 / np.array([1, 2, 0, 5])
```
