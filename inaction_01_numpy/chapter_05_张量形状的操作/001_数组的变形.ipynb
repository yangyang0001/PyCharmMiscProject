## 数组的变形、转置相关操作

### 1. 数组的变形

```python
import numpy as np

a = np.arange(10, 100, 2)
a
```

```python
a.size
```

```python
# 是浅拷贝; 共用一份数据
b = a.reshape(9, 5)
b
```

```python
b[0][0] = 100
b
```

```python
a
```

### 2. 数组的转置

```python
a = np.random.randint(0, 100, size = (10, 3))
a
```

```python
# 张量的转置; 或者用 np.transpose 方法
#a.T
a.transpose()
```

```python
a
```

```python
np.transpose(a, (1, 0))
```

```python
b = np.random.randint(0, 10, size = (3, 5, 7))
b.shape
```

### 3. 数组的堆叠

```python
a = np.random.randint(0, 10, size = (3, 5))
b = np.random.randint(0, 10, size = (3, 5))

# 增加行的操作; 默认合并 行增加 axis 有轴的意思; 0:表示横向或水平方向 上堆叠 为默认值; 1:表示 垂直方向上 堆叠
np.concatenate((a, b), axis = 1)
```

```python
# 水平方向上进行堆叠；列增加
np.hstack((a, b))
```

```python
# 垂直方向上进行堆叠；行增加
np.vstack((a, b))
```

```python
np.vstack((a,a,b))
```

### 4. 数组的拆分

```python
a = np.random.randint(0, 100, size = (15, 10))
a
```

```python
# 如果第二个必填参数为数字; 则表示行 平均分成几份
np.split(a, indices_or_sections=3)
```

```python
np.split(a, indices_or_sections=2, axis=1)
```

```python
# 如果indices_or_sections 给出一个数组; 这里会按照具体的点进行切分
np.split(a, indices_or_sections=[1, 3, 7, 9])
```

```python
# 有水平堆叠和垂直堆叠; 就有 水平方向拆分和垂直方向上拆分
np.hsplit(a, indices_or_sections=2)
```

```python
np.vsplit(a, indices_or_sections=[5])
```
