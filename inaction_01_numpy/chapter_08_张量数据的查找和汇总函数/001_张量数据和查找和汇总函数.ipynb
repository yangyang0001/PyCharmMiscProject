## 张量的where sort 操作和统计函数

### 1. 张量的 where 函数

```python
import numpy as np

nd1 = np.array([1, 2, 3, 4, 5, 6])
nd2 = np.array([10, 20, 30, 40, 50, 60])
display(nd1, nd2)
```

```python
# where函数 如果 True 返回 nd1 中元素; False 返回 nd2 中的元素
cond = np.array([True, False, True, True, False, False])
np.where(cond, nd1, nd2)
```

```python
a = np.random.randint(0, 100, size = 50)
a
```

```python
# where 函数; 大于50 则展示a中的数据; 否则展示a+20; 后面的随便填写就OK了
np.where(a > 50, a, a+20)
```

```python
a = np.random.randint(0, 100, size = 50)
# 从小到大进行排序; 没有输出, 就是改的原始数据
a.sort() 
a
```

```python
cond = (a > 50) & (a < 60)
np.where(cond, a + 10, a)
```

### 2. sort 和 argsort 函数

```python
a = np.random.randint(0, 100, size = 50)
a
# 获取排序后 原始数据的下标 张量
index = a.argsort()
index
# 根据花式索引 根据下标 index 获取 从小到大的数据
a[index]

# 根据花式索引获取 从大到小的数据; 两种获取方式
display(a[index][::-1], a[index[::-1]])
```

### 3. 集合操作函数

```python
a = np.random.randint(0, 100, size = 30)
b = np.random.randint(0, 100, size = 30)

display(a, b)
# 交集
display(np.intersect1d(a, b))
# 并集
display(np.union1d(a, b))
# 差集
display(np.setdiff1d(a, b))
```

### 4. 数据和统计或汇总函数

```python
a = np.random.randint(10, 100, size = (3, 5))
a
```

```python
# 返回 符合条件的 索引
index = np.argwhere(a > 40)
```

```python
for i,j in index:
    print(a[i, j])
```

```python
# 协方差 (两个属性之间进行计算) 方差: （数据内部, 属性内部计算)
# 举例: 一个男人的受欢迎程度 和 一个男人的语言表达水平有多大的关系
np.cov(a)
```

```python
# 相关性系数; 和协方差有关 0:无关 1:
np.corrcoef(a)
```
