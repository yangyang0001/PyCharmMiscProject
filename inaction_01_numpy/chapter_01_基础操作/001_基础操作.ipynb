## 基本操作

### 1. 数组的创建

```python
import numpy as np

# 创建一个数组
l = [1, 2, 3, 4, 5]
nd1 = np.array(l) # 输入 arr + tab 键可以自动补全
print(nd1)
display(nd1)
```

```python
nd2 = np.zeros(shape = (3, 4), dtype = np.int32)
nd2
```

```python
nd3 = np.ones(shape=(3, 4), dtype=np.float32)
nd3
```

```python
nd4 = np.full(shape= (2, 3, 4), fill_value=3.1415926) # 生成任意形状的 数组
nd4
```

```python
nd5 = np.random.randint(0, 100, size = 15) # 生成 0 到 100 之间的一个 长度为 20的数组
nd5
```

```python
nd6  = np.random.rand(2, 3) # 生成 2行3列的 数组; 填充值为 0到1
nd6
```

```python
nd7 = np.random.randn(3, 5) # 生成一个 正态分布 数组; 所有数的 理论上的平均值为 0 标准差为1
nd7
```

```python
nd8 = np.random.normal(loc = 100, scale = 10, size = (1, 5)) # 生成一个 1行5列的 正态分布；平均值为 100，标准差为 10
nd8
```

```python
nd9 = np.arange(1, 100) # 等差数列 差值为1；1 到 100 的一个一维数组
nd9
```

```python
nd91 = np.arange(1, 100, step = 10) # 等差数列 差值为 10; 左闭右开 的一个等差数列
nd91
```

```python
nd10 = np.linspace(1, 100, num = 10) # 等差数列 差值为 10; 左闭右闭 的一个等差数列；数字个数为 10 个
nd10
```

### 3. 文件操作

```python
import numpy as np

nd1 = np.random.randint(0, 10, size=5)
nd2 = np.random.randn(2, 4)
display(nd1, nd2)
```

```python
# 把 单个数组数据保存到文件中; 只保存路径就可以了, 文件自动保存为 后缀为 .npy 的文件
np.save('./data', nd2)
```

```python
# 加载 单个文件数据
np.load('./data.npy') 
```

```python
# 把多个文件 保存在一个文件中; a, b 等我们称之为 key
np.savez('./data', a = nd1, b = nd2)
```

```python
# 加载 单个文件数据; 数据获取必须 要加上 key
data = np.load('./data.npz')
data
```

```python
# 加载 单个文件数据; 数据获取必须 要加上 key
data['a']
```

```python
# 加载 单个文件数据; 数据获取必须 要加上 key
data['b']
```

```python
# 如果获取的不是 数据中有的key 直接报错
data['wwww']
```

```python
# 保存压缩文件
np.savez_compressed('./data2.npz', x = nd1, y = nd2)
```

```python
data2 = np.load('./data2.npz')
data2
```

```python
data2['x']
```

```python
data2['y']
```

```python
# 保存 txt 文件
np.savetxt(fname= './data.txt', X = nd1, fmt='%0.2f', delimiter=',')
```

```python
np.loadtxt('./data.txt')
```

```python
# 保存 csv 文件
np.savetxt(fname='./data.csv', X = nd1, fmt='%d', delimiter=';')
```

```python
np.loadtxt('./data.csv')
```
