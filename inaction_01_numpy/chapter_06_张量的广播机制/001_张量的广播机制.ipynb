## 张量的广播机制

```python
import numpy as np

a = np.arange(20)
a
```

```python
b = a.reshape(4, 5)
b
```

```python
arr1 = np.array([0, 1, 2, 3] * 3)
arr1.sort()
arr1 = arr1.reshape(4, 3)
arr2 = np.array([1, 2, 3])
display(arr1, arr2)
```

```python
# arr1 和 arr2 虽然形状不一样, 仍然能够进行计算, 这是 numpy 对张量进行了广播所致; 行不够 广播行
arr1 + arr2
```

```python
# arr1 和 arr3 虽然形状不一样, 仍然能够计算, 列不够, 广播列
arr3 = np.array([[1], [2], [3], [4]])
arr3
```

```python
arr1 + arr3
```

```python
display(arr1.shape, arr3.shape)
```

```python
a = np.array([1, 2, 3, 4, 5, 6, 7, 8] * 3).reshape(3, 4, 2)
a
```

```python
b = np.array([0, 1, 2, 3, 4, 5, 6, 7]).reshape(4, 2)
b
```

```python
a + b
```
