# 遍历和迭代

```python
a = list(range(10))

for num in a:
    print(num)
```

```python
b = {'a': 10, 'b': 20, 'c': 100}
for key in b:
    print(b.get(key))
```

```python
c = [(1, 2), (2, 3)]
for x, y in c:
    print(x, y)
```

## 结论: python 中的遍历和迭代比java 要抽象的多, 底层实现极其强大

```python
def findMaxAndMinNum(array):
    maxNum = minNum = None    # python 没有 Long.MIN_VALUE, 用 None 表示还没有值
    for num in array:
        if maxNum is None or num > maxNum:
            maxNum = num
        if minNum is None or num < minNum:
            minNum = num
    return (minNum, maxNum)

print(findMaxAndMinNum(range(10)))
print(findMaxAndMinNum([7, -3, 15, 0]))
print(findMaxAndMinNum([]))
```
