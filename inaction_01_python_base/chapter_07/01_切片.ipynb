## 编写一个 trim 方法

```python
def trim(s):
    start, end = 0, len(s)
    while start < end and s[start] == ' ':   # 从左往右，跳过开头的空格
        start += 1
    while end > start and s[end - 1] == ' ':  # 从右往左，跳过结尾的空格
        end -= 1
    return s[start:end]                       # 只切片一次
print(trim('zhang  san'))
```

### 优化: 只去掉首尾空格, 用下标定位后切片一次
- 上面的写法会删掉所有空格, `'zhang  san'` 变成 `'zhangsan'`
- 下标只移动不复制字符串, 最后切片一次, O(n)
- `start < end` 保证 `''` 和 `'    '` 不会越界

```python
def trim(s):
    start, end = 0, len(s)
    while start < end and s[start] == ' ':   # 从左往右，跳过开头的空格
        start += 1
    while end > start and s[end - 1] == ' ':  # 从右往左，跳过结尾的空格
        end -= 1
    return s[start:end]                       # 只切片一次

print(len(trim('   123 5 ')))
print(trim('   123 5 '))
```
