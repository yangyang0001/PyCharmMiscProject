## 数组 和 Tuple

### 数组

```python
# 使用数组
a = ['zhangsan', 'lisi', 'wangwu']
print(a)
len(a)

# 正向获取
print(a[0], a[1], a[2])

# 从 index 返向获取
print(a[-1])

# 数组 末尾添加元素
a.append('zhaoliu')
print(a)

# 数组指定位置 插入元素
a.insert(1, 'gouzi')
print(a)

# 删除指定位置的 元素
a.pop(1)
print(a)
```

```python
# 在 python 中 数组是可以存放不同类型的数据的
L = ['zhangsan', 20, True]
print(L)
```

```python
# list 元素可以是另外一个 数组
arr = ['zhangsan', 'lisi', ['gouzi', 'xiaomao']]
print(len(arr))
print(arr[2])
```

### tuple 元组

```python
# tuple 定一个用 (); 和 list 使用[] 最大区别是 一旦定义 tuple 内部元素不能被改变
t = (1, 2, 3)
print(t)
print(t[1])

a = ['zhangsan', True]
t = (1, 2, a)
print(t)
a[1] = 'lisi'
print(t)
```
