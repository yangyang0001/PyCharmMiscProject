## 字典 (map) 和 Set

### dict 字典或着 map 或者 json

```python
# 存放学生成绩的 dict 就是 map; 也可以认为是 json
d = {'zhangsan': 85.5, 'lisi': 22.5, 'wangwu': 88}
print(d['wangwu'])

# 修改 dict 中的数值
d['wangwu'] = 100
print(d['wangwu'])
print(d)
```

```python
# 判定 key 是否存在 使用格式 key in dictName
d = {'zhangsan': 'hello'}
print(d['zhangsan'])
print('wangwu' in d)
```

```python
# 获取 key 的值; 如果不存在 可以设置默认值
d = {'zhangsan': 'hello'}
# 如果没有获取到 zhangsan1 则给出默认值 123
print(d.get('zhangsan1', '123'))
```

```python
# dict 中的添加元素和删除元素
d = {'zhangsan': 10, 'lisi': 20}

# 添加元素
d['wangwu'] = 100

# 使用 update 添加元素 或者 修改元素
d.update({'zhaoliu': 120})
d.update({'zhaoliu': 88, 'maosan': 99})
print(d)

# 删除元素
d.pop('wangwu')
print(d)
```

### Set

```python
# 构造一个 set 也是用 {} 来表示
s = {1, 2, 3}
print(s)
```

```python
# 使用 list 作为参数构造集合; set 是去重集合
s = set([10, 10, 20, 30, 30])
print(s)
s.add(5.5)
print(s)

# set 添加元素; (有 0 和 1 的 set True 当作1; False 当作0)
s = {0, 1, 3, 5}
s.add(False)
s.add(True)
print(s)

# set 删除元素
s.remove(1)
print(s)
```

```python
# 既然 set 类似于 数学中的集合 那么集合 必然符合数学中的集合操作
s1 = {1, 2, 3, 4, 5}
s2 = {1, 3, 5, 7, 9}
# 集合的 交集 并集 补集
print(s1 & s2)
print(s1 | s2)
print(s1 - s2)
```

### 再谈 不可变对象

```python
a = ['b', 'c', 'a']
a.sort()
print(a)
```

```python
a = 'abc'
a.replace('a', 'A')
print(a) # a的结果仍为 'abc'; 证明 a 是不可变字符串
```

```python
### set 中塞入不同的数据集合; 这里塞入的是 tuple 不可变元组
a = (1, 2, 3, 3, 2, 1)
b = set(a)
print(b)

a = (1, 2, [3, 4])
print(a)
b = set(a)
print(b)
```
