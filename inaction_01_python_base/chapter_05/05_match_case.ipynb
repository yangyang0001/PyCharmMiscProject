```python
## 类似于 java 中的 case 语句; _表示任何字符; 这是 python 
score = 'aa'
match score:
    case 'A':
        print('A')
    case 'B':
        print('B')
    case _:
        print('其他字符')
```

```python
# 复杂匹配; 赋值给一个变量; 让变量去匹配
age = 35
match age:
    case x if x < 10:
        print('x is %s' % x)
    case x if 10 <= x <= 18:
        print('age is 10 - 18')
    case 19 | 20 | 21 | 22 | 23 | 24:
        print('age is 19 到 24')
    case _:
        print('其他岁数')
```
