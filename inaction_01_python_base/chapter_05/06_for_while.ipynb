## 循环 和 break continue操作

```python
# for 循环
sum = 0
for x in range(1, 11):
    sum += x
print(sum)    
```

```python
# while 循环
n = 1
sum = 0
while n < 11:
    sum += n
    n += 1
print(sum)    
```

```python
# break 操作
sum = 0
for x in range(1, 11):
    sum += x
    if sum > 20:
        print('x = %s' % x)
        break
print(sum)        
```

```python
# continue 操作
sum = 0
for x in range(1, 11):
    if x % 2 == 0:
        continue
    sum += x
print(sum)    
```
