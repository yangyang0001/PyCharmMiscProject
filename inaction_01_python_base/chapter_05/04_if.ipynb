## 条件判断

```python
# 通常使用的 if-else 使用方式; 有: 
age = 20
if age > 20:
    print('your age is %s' % age)
else:
    print('age less than 20')
```

```python
# 使用的 if-elif-else 使用方式 都有:
age = 100
if age > 20:
    print('age is great than %s' % 20)
elif age >= 10:
    print('age is great eqal %s' % 10)
else:
    print('age less then 10')
```

```python
# 非None 非0 等都为 True 的判定
x = None
if x:
    print(True)
else:
    print(False)
```

```python
# 通过输入判定
birth = input('input your birth')
birth = int(birth)
if birth > 2000:
    print('00后')
else:
    print('00前')
```

```python
# 一定要记住 : 的使用方式
## 多重 if elif elif elfi else
score = 75.5
if score > 90:
    print('优秀')
elif score > 80:
    print('良好')
elif score > 60:
    print('及格')
else:
    print('不及格')
```
