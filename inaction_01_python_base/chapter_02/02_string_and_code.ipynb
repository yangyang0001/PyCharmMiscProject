```python
# 直接打印出 ASCII 码
print(ord('A'))
print(ord('中'))


print('你好'.encode('utf-8'))                        # str -> bytes
b'\xe4\xbd\xa0\xe5\xa5\xbd'.decode('utf-8')     # bytes -> str
```

```python
# ord 只能针对一个 字符; 针对 ord 有 chr 这是一对
print(ord('中'))
print(chr(20013))
```

```python
print(ord('杨'))
print(chr(ord('杨')))
```

```python
# encode 和 decode 是一对
a = '中华'
print(a.encode('utf-8'))
print(b'\xe4\xb8\xad\xe5\x8d\x8e'.decode())
```

```python
# 计算字符串的长度
print(len('ABC'))
print('中华')
```

```python
# %? 表示替换符；格式为 [%? %]只有一个替换符 () 可以省略
print('Hello %s' % '张三')
print('%2d-%02d' % (3, 1))
print('%.2f' % 3.1415926)
# 如果不确定用什么 %s 永远好用
print('Age: %s Gender: %s' % (25, True))
# 如果表示单个字符串的情况 用 % 来修饰
print('%2d%%' % 10)
```

```python
# f-string 表示替换符
r = 2.5
s = 3.14 * r * r
print(f'{r} area is {s:.2f}')
```

```python
a = 72
b = 85
r = (b - a) / 85
print(f'{r:2f}')
```
