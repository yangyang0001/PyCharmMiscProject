# 递归函数的使用

## 定义一个递归函数

```python
def fact(n):
    if n == 1:
        return 1
    return n * fact(n - 1)

print('5! =', fact(5))
```

### 注意点: 栈帧不是无限的, 要注意压栈的层数

```python
print('10! =',fact(10))
```

## 编写一个函数 move(n, a, b, c)函数，它接收参数n，表示3个柱子A、B、C中第1个柱子A的盘子数量，然后打印出把所有盘子从A借助B移动到C的方法

```python
## 1个盘子:   a ---> c
## 2个盘子:   a ---> b    a ---> c   b ---> c
## 3个盘子:   a ---> c    a ---> b   c ---> b   a ---> c  然后 将 b 柱子当成 a 柱子，然后重复 两个盘子的方式

# 思路: 把 n 个盘子从 a 移到 c (b 作为中转)
#   1. 先把上面 n-1 个盘子从 a 移到 b (c 作为中转)
#   2. 把最大的第 n 个盘子从 a 移到 c
#   3. 再把 b 上的 n-1 个盘子移到 c (a 作为中转)
# 递归出口: n == 1 时直接 a ---> c
def move(n, a, b, c):
    if n == 1:
        print(a, '--->', c)
        return
    move(n - 1, a, c, b)
    print(a, '--->', c)
    move(n - 1, b, a, c)

move(5, 'A', 'B', 'C')
```
