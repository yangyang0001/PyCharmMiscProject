## 定义函数

```python
from sympy.integrals import quadrature


# 定义函数 规则 def 函数名(参数列表); 如果没有返回值 默认为 None
def my_abs(x):
    if x > 0:
        return x
    else:
        return -x

print(my_abs(-11))
```

```python
import math

def quadratic(a, b, c):
      delta = b*b - 4*a*c
      sqrt_delta = math.sqrt(delta)
      # 返回两个 根
      x1 = (-b + sqrt_delta) / (2*a)
      x2 = (-b - sqrt_delta) / (2*a)
      return x1, x2

print(quadratic(1, 4, 3))
```

## 定义一个返回多个值的函数

```python
def mine(x, y):
    return x, y
(a, b) = mine(10, 20)
print(a, b)
```
