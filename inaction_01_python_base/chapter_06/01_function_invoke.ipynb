## 函数调用

```python
#  取绝对值函数
print(abs(100))
print(abs(-20))
print(abs(12.34))
```

```python
# 取最大值 最小值函数
print(max(1, 3, 5))
print(min(1, 2, 3))
```

```python
# 字符串 ---> 数字
print(int('123'))
print(float('123'))

# 数字 ---> 字符串
print(str(123))

# boolean 函数
print(bool(1))
print(bool(''))
```

```python
# 可以给函数起别名; 即 函数可以 赋值给 变量; 使用规则: 变量名 = 函数名
a = abs
print(a(-1))
```

```python
# 练习 数字 ---> 16进制
n1 = 1000
n2 = 1024
print(hex(n1))
print(hex(n2))
```
