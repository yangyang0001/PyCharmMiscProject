# 函数的参数

````python
def power(x):
    return x * x

print(power(2))
````

````python
#假设现在要求 x的 n次方
import math
def mine_power(x, n=2):
    return pow(x, n)

print(mine_power(2, 5))
````

````python
# 可变参数 和 关键字参数 和 命名关键字参数
def mine_param(a, b, *param0, param1, param2, **param3):
    print(a)
    print(b)
    print(param0)
    print(param1)
    print(param2)
    print(param3)
param1 = {'param2':'param2'}
param2 = {'param3':'param3'}

# 必须如下的方式进行调用
mine_param(1, 2, [1,2,3], param1=param1, param2=param2)
````

## 参数规则总结

### 一、定义函数时的规则

参数顺序（固定，不能打乱）：

```python
def f(a, b, /, c, d=10, *args, e, f=20, **kwargs):
#     └──┬──┘  └──┬───┘  └─┬─┘ └──┬───┘  └──┬───┘
#   仅位置参数  普通参数   可变参数 命名关键字参数 关键字参数
```

| 顺序 | 类型 | 写法 | 说明 |
|---|---|---|---|
| 1 | 仅位置参数 | `a, b, /` | `/` 前面的参数只能按位置传（Python 3.8+，了解即可） |
| 2 | 普通参数 | `c, d=10` | 既可按位置传，也可按名字传 |
| 3 | 可变参数 | `*args` | 收集多余的位置参数，得到 **元组** |
| 3' | 分隔符 | `*` | 没有 `*args`、又需要命名关键字参数时用 |
| 4 | 命名关键字参数 | `e, f=20` | 必须写 `名字=值` 传 |
| 5 | 关键字参数 | `**kwargs` | 收集多余的关键字参数，得到 **字典**，**必须放最后** |

1. **`**kwargs` 必须是最后一个参数。**
2. **`*args` 和单独的 `*` 只能二选一。** 有了 `*args`，它后面的参数自动就是命名关键字参数。
3. **单独的 `*` 后面必须至少有一个命名关键字参数。** `def f(a, *):` 会报错。
4. **在 `*` 之前，有默认值的参数要放在没有默认值的参数后面。** ❌ `def f(a=1, b):`，✅ `def f(b, a=1):`
5. **命名关键字参数不受第 4 条限制。** `def f(*, e=1, g):` 是合法的。
6. **`*args` 和 `**kwargs` 各自最多出现一次。**
7. **默认值不要用可变对象**（`[]`、`{}`），否则多次调用会共用同一个对象，应该用 `None`。

### 二、调用函数时的规则

```python
f(1, 2, *[3, 4], c=5, **{'d': 6})
#  └─┬─┘ └──┬──┘  └─┬┘ └───┬───┘
# 位置参数 位置解包 关键字参数 字典解包
```

1. **位置参数必须在关键字参数前面。** ❌ `f(a=1, 2)`，❌ `f(**d, 2)`
2. **调用时的 `*` 和 `**` 是"解包"。** 后面必须跟对象：`*` 跟列表、元组等，`**` 跟字典。**不能单独写 `*`。**
3. **命名关键字参数必须写 `名字=值`。** 变量名和参数名相同也没用，Python 不会自动对应。
4. **同一个参数只能赋值一次。** `def f(a):` 用 `f(1, a=2)` 调用会报 `got multiple values for argument 'a'`。
5. **`**` 解包的字典，键必须是字符串，而且不能和其他关键字参数重名。**
6. **关键字参数之间、关键字参数和 `**解包` 之间的先后顺序随意。**
7. **例外：`*解包` 可以出现在 `名字=值` 后面，但不能出现在 `**解包` 后面。** `g(b=3, *[1, 2])` 合法，`g(**d, *[2])` 报错。这种写法可读性差，实际还是按"位置 → `*解包` → 关键字 → `**解包`"来写。

### 三、Python 匹配参数的过程

1. 位置参数按顺序填进形参。
2. 多出来的位置参数交给 `*args`；如果没有 `*args`，报错。
3. 关键字参数按名字匹配形参。
4. 名字对不上的关键字参数交给 `**kwargs`；如果没有 `**kwargs`，报错。
5. 还没赋值的参数用默认值；没有默认值就报 `missing required argument`。

**一句话记忆：** 定义时按"位置 → `*` → 命名关键字 → `**`"排；调用时"不带名字的在前，带名字的在后"。

````python
# 规则示例
def mine_param(a, b, *param0, param2, param3='默认', **param1):
    print(a, b, param0, param2, param3, param1)

mine_param(1, 2, 3, 4, param2='p2', name='zhangsan')
# 1 2 (3, 4) p2 默认 {'name': 'zhangsan'}
````
